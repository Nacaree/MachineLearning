```python
import numpy as np # deal array one or two dimensions
import pandas as pd # import data and virtual as table
import matplotlib.pyplot as plt # vitual data as graph

# Convert Independent variables to Polynomial Features
# Polynomial Math following degree
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

#step 2: load data and virtual data in the table 
dataset = pd.read_csv("salary.csv");
print(dataset);

#step 3: data preprocessing or data engineering
#input or independence variable
X = dataset[["Level"]].values;
#output or dependent variable
Y = dataset["Salary"].values;

#step 4: covert input variable to polynomial Features
polynomial_features = PolynomialFeatures(degree=4)
#transform X = [1,X,x^2, X^3, X^4] to give curve line
#fit -> polynomial learn x
#transform 0> convert input variable to polynomial math
X_poly = polynomial_features.fit_transform(X);
print("==== Polynomial Features ====");
print(X_poly);

#step 5: create and train Polynomial Regression Model
polynomial_model = LinearRegression();
polynomial_model.fit(X_poly, Y);

#Step 6: predict salary for level = 6.5
position = np.array([[6.5]])
#convert 6.5 to polynomial features
position_poly = polynomial_features.transform(position);
#conevrt 6.5 exprience level, how much he got
position_predict = polynomial_model.predict(position_poly);
print("6.5 level of Expererience", position_predict[0], "$");
```
