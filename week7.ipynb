```python
import pandas as pd
# pandas got 2 types : Series, and Data Frame 
# pandas series is an enhanced array one size
grades = pd.Series([80,85,90,95], index=["A","B","C","D"])
print(grades)
print("\n=== data manipulation ===")
print(grades["B"])
print(grades["D"])

print("Add value to Pandas Series")
grades["E"] = 100
print(grades)

print("Delete Specific index")
grades = grades.drop("B")
print(grades)

print("Descriptive statistics")
print("Count: ", grades.count())
print("Min: ", grades.min())
print("Max:: ", grades.max())
print("Mean: ", grades.mean())
print("Std: ", grades.std())

print("Descriptive statistics")
print(grades.describe())
```

```python
# numpy is n-dimensional array
# from numpy you can create 1, 2 n-dimensional array
import numpy as np
array1 = np.array([10, 20, 30, 40, 50])

print(array1)
print(array1[3])
# add new value to numpy array
array1 = np.append(array1, 60)
print(array1)
# delete specific index
array1 = np.delete(array1, 2)
print(array1)
```

```python
array2 = np.array([
    [1,2,3,4,5],
    [10,20,30,40,50],
    [100,200,300,400,500]
])

print("sum: ", array2.sum()) #* sum of all numbers in array
print("min: ", array2.min()) #* lowest number in the array
print("max: ", array2.max()) #* highest number 
print("mean: ", array2.mean()) #* average from the sum of array
print("std: ", array2.std())  #* how spread out the numbers are from an axis
```
