```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

data = {
    'House Size (m²)': np.array([50, 70, 90, 110, 130, 150, 170, 190]),
    'Price ($1000)': np.array([120, 150, 185, 220, 250, 285, 310, 350])
}

df = pd.DataFrame(data)
print(df)

# Separate X and Y
x = df[['House Size (m²)']]
y = df['Price ($1000)']

# Create and train the model
model = LinearRegression()
model.fit(x, y)

# Extract slope and intercept
m = model.coef_[0]
b = model.intercept_

# Predict the price for 125 m²
target_house = pd.DataFrame({'House Size (m²)': [125]})
predicted_price = model.predict(target_house)[0]

print(f"Slope: {m:.6f}")
print(f"Intercept: {b:.6f}")
print(f"Equation: y = {m:.6f}x + {b:.6f}")
print(f"Predicted price: ${predicted_price * 1000:,.2f}")
# * Visualize data in scatter graph
plt.scatter(x, y, color = 'blue', label = 'Actual prices')
plt.plot(x, model.predict(x), color = 'red', label='Regression line')
plt.scatter(125, predicted_price, color = 'green',
            s = 80, label = '125 m² prediction')

plt.xlabel('House Size (m²)')
plt.ylabel('Price ($1000)')
plt.title('House Size vs Price')
plt.legend()
plt.show()
```
