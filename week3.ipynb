```python
# Conditional Statement 2 types 
salary = int(input("Enter Salary "))
if salary > 100:
    print("Salary",salary)
    print("Continue body of if")
elif salary > 50:
    print("salary is greater than 50 and lower 50", salary)
else:
    print("else", salary)

# Switch Statement for fit options / values
match salary:
    case 100:
        print("Salary is 100")
    case 200:
        print("Salary is 200")
    case _:
        print("else of above condition")
    
```
