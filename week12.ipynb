```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

# Create a DataFrame with the pizza menu data
data = {'Diameter (inches)': [6, 8, 10, 14, 18],
        'Price (dollars)': [7, 9, 13, 17.5, 18]}
# convert data to table 
df = pd.DataFrame(data)
print(df)
# step 3: data processing and feature engineering
# get summary data
print("==== Summary data frame ====")
print(df.describe())
# check null value of data frame
print("==== check null value ====")
print(df.isnull().sum())
# visual data in a scatter graph
plt.scatter(df['Diameter (inches)'], df['Price (dollars)'])
plt.xlabel('Diameter (inches)')
plt.ylabel('Price (dollars)')
plt.title('Pizza Menu')
plt.show()
# * step 4: split data to X (independent variable) and Y (dependent variable)
x = df[['Diameter (inches)']]
y = df['Price (dollars)']
# * step 5: create LinearRegression algorithms and model
model = LinearRegression()
model.fit(x, y) # ? train LinearRegression ALGO to create LinearRegression ML 
# Extract slop (m) and intercept (b)
m = model.coef_[0]
b = model.intercept_
```
