```python
import statistics

sample = [10,20,30,30,40,50, 1000, 9000]
print("Min: ", min(sample))
print("Max: ", max(sample))
print("Count: ", len(sample))
print("======= Measuer of central =======")
print("Mean: ", statistics.mean(sample))
print("Median: ", statistics.median(sample))
print("Mode: ", statistics.mode(sample))
```

```python
import statistics 
number_list = []

def enter_numbers(number_list):
    numbers = int(input("Enter numbers please: "))
    number_list.append(numbers)
    
def display_numbers(number_list):
    print("Numbers:", number_list)
    
def print_mean(number_list):
    print("Mean: ", statistics.mean(number_list))
    
def print_median(number_list):
    print("Median: ", statistics.median(number_list))
    
def print_mode(number_list):
    print("Mode: ", statistics.mode(number_list))

def print_std_dev(number_list):
    print("Standard Deviation: ", statistics.stdev(number_list))

while True:
    print("====== MENU ======")
    print("1. Enter Numbers")
    print("2. Display Numbers")
    print("3. Calculate Mean")
    print("4. Calculate Median")
    print("5. Calculate Mode")
    print("6. Calculate Standard Deviation")
    print("7. Exit")
    choice = input("Choose an option (1-7): ")
    
    if choice == "1":
            enter_numbers(number_list)
    elif choice == "2":
            display_numbers(number_list)
    elif choice == "3":
            print_mean(number_list)
    elif choice == "4":
            print_median(number_list)
    elif choice == "5":
            print_mode(number_list)
    elif choice == "6":
            print_std_dev(number_list)
    elif choice == "7":
            print("Exiting...")
            break
    else:
            print("Invalid choice. Please select 1-7.")

```
