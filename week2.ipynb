```python
price = 0
total = 0
while True:
    print("==== Menu ====")
    print("1. Burger $5")
    print("2. Pizza $8")
    print("3. Pasta $7")
    print("4. Salad $4")
    print("5. exit")
    
    name = int(input("Enter Menu Number: "))
    age = int(input("Enter qty: "))
    
    if  name == 5:
        break
    elif name == 1:
        price = 5
        print("Burger")
        print("Quanity: ", age)
    elif name == 2:
        price = 8
        print("Pizza")
        print("Quanity: ", age)
    elif name == 3:
        price = 7
        print("Pasta")
        print("Quanity: ", age)
    elif name == 4:
        price = 4
        print("Salad")
        print("Quanity: ", age)
    else:
        print("invalid")
        break

    discounted = (price * 5/ 100)
    total = price - discounted
    print("Quantity: ", age)
    print("Discount: 5%")
    print("Total Cost: ", total )
   
```

```python
balance = 1000
while True:
    print("==== Menu ====")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")
    option = int(input("Choose: "))
    
    print("==== Output ====")
    match option:
        case 1:
            print("Balance:",balance)
        case 2:
            deposit = int(input("Deposit Amount: "))
            balance = balance + deposit
            print("New Balance:", balance)
        case 3:
            withdraw = int(input("Withdraw Amount: "))
            balance = balance - withdraw
            print("New Balance :", balance)
        case 4:
            print("BYEEEEEEEEE")
            break
    
```
