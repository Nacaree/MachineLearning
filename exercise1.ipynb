```python
id = input("Enter ID: ")
name = input("Enter Name: ")
age = input("Enter Age: ")
Major = input("Enter Major: ")
Score = []
for i in range(0, 3): 
    add = float(input("Enter Score: "))
    Score.append(add)

Average = sum(Score) / 3 
grade = ""
if Average >= 90:
    grade = "A"
elif Average >= 80:
    grade = "B"
else:
    grade = "F"    
# Average = Score / 3
print("=== Output ===")
print("Student ID:",id)
print("Name:",name)
print("Age:",age)
print("Major:",Major)

print("-- Score Result --")
print("JavaScript:",Score[0])
print("Database:",Score[1])
print("Networking:",Score[2])
print(f"Average: {Average:.2f}")
print("Grade: ",grade)
```
